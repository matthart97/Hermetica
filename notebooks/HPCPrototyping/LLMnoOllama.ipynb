```python
import requests
import os

# Replace these with your values
repo_id = "TheBloke/TinyLlama-1.1B-Chat-v1.0-GGUF"
filename = "tinyllama-1.1b-chat-v1.0.Q4_K_M.gguf"
output_dir = "./models"
#token = "YOUR_HF_TOKEN"  # Replace with your actual token

# Create URL for the file
url = f"https://huggingface.co/{repo_id}/resolve/main/{filename}"

# Setup authentication
headers = {"Authorization": f"Bearer {token}"}

# Create output directory
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, filename)

# Download the file
print(f"Downloading {filename}...")
response = requests.get(url, headers=headers)

# Check if successful
if response.status_code == 200:
    # Save the file
    with open(output_path, 'wb') as f:
        f.write(response.content)
    print(f"Downloaded to {output_path}")
else:
    print(f"Error: {response.status_code}")
    print(f"Message: {response.text}")
```

```python
from llama_cpp import Llama

# Path to your downloaded model
model_path = "./models/tinyllama-1.1b-chat-v1.0.Q4_K_M.gguf"  # Change to your model path

# Initialize the model
llm = Llama(
    model_path=model_path,
    n_ctx=2048,         # Context window size
    n_gpu_layers=-1     # Use GPU if available (-1 means all layers)
)

# Define a simple prompt
prompt = "<|user|> What is the capital of France? <|assistant|>"

# Generate a response
print(f"Prompt: {prompt}\n")
print("Generating response...\n")

output = llm(
    prompt,
    max_tokens=100,    # Generate up to 100 tokens
    temperature=0.7    # Controls randomness (0.0 to 1.0)
)

# Print the generated text
print("Response:")
print(output["choices"][0]["text"])
```

```python
from llama_cpp import Llama

# Path to your downloaded model
model_path = "./models/tinyllama-1.1b-chat-v1.0.Q4_K_M.gguf"  # Change to your model path

# Initialize the model
llm = Llama(
    model_path=model_path,
    n_ctx=2048,         # Context window size
    n_gpu_layers=-1     # Use GPU if available (-1 means all layers)
)

# Inspect model metadata and parameters
print("=== MODEL INFORMATION ===")
print(f"Model path: {model_path}")
print(f"Context size (n_ctx): {llm.n_ctx()}")
print(f"Vocabulary size: {llm.n_vocab()}")
print(f"Model size: {llm.n_embd()} dimensions")
print(f"Model layers: {llm.n_layer()}")

# Get model metadata from GGUF file
print("\n=== MODEL METADATA ===")
if hasattr(llm, 'metadata') and llm.metadata:
    for key, value in llm.metadata.items():
        print(f"{key}: {value}")
else:
    print("Metadata not directly accessible through this interface")

# Check if the model has chat template information
print("\n=== CHAT TEMPLATE INFO ===")
try:
    if hasattr(llm, 'metadata') and 'tokenizer.chat_template' in llm.metadata:
        print(f"Chat template: {llm.metadata['tokenizer.chat_template']}")
    else:
        print("No specific chat template found in metadata")
except:
    print("Could not access chat template information")

# Print tokenizer details
print("\n=== TOKENIZER SAMPLE ===")
sample_text = "Hello, world!"
tokens = llm.tokenize(bytes(sample_text, 'utf-8'))
print(f"Sample text: '{sample_text}'")
print(f"Tokenized to {len(tokens)} tokens: {tokens}")

# Print sample special tokens if available
print("\n=== SPECIAL TOKENS ===")
special_tokens = {
    "BOS": llm.token_bos(),
    "EOS": llm.token_eos(),
    "NL": llm.token_nl()
}
for name, token_id in special_tokens.items():
    print(f"{name} token: {token_id}")
```
